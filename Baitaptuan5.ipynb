```python
#W5A1
#Viết hàm truyền vào hai số nguyên a, b. Trả về số lớn hơn trong hai số. def Max_Of_Two(a, b) -> int:
def Max_Of_Two(a, b):
    return max(a, b)
a, b = map(int, input("").split('/'))
print(f'So lon hon la: {Max_Of_Two(a,b)}')
```

```python
#W5A2
#Viết hàm hoán đổi hai số nguyên. def swap(a, b) -> Tuple[int, int]:
def swap(a,b):
    a = b + a
    b = a - b
    a = a - b
    return [a, b]
a, b = map(int, input().split())
swap(a,b)
```

```python
#W5A3 
#Viết hàm kiểm tra một số có phải là số nguyên tố hay không? 
def prime_check(n):
    if n >= 2:
        check = True
        for i in range (2, int(n**(1/2))+1):
            if n % i == 0:
                check = False
                break
        if check:
            print (f'{n} la so nguyen to')
        else:
            print (f'{n} khong phai la so nguyen to')
    else:
        print (f'{n} khong phai la so nguyen to')
while True:
    try:
        n = int(input())
        prime_check(n)
        print ("Ban muon lam lai khong?")
        m = input()
        if m == "yes":
            continue
        else:
            print ("Tam biet")
            break
    except ValueError:
        print ("Nhap lai")
```

```python
#W5A4
#VIết hàm kiểm tra một số có phải số hoàn hảo hay không, xử lý trên dữ liệu kiểu số nguyên
def perfect_number(n):
    if n == 0:
        return False
    tong = 0
    for i in range (1, n//2 + 1):
        if n % i == 0:
            tong += i
    if n == tong:
        return True
    else:
        return False
n = int(input())
m = perfect_number(n)
if m == True:
    print (f'{n} is a perfect number')
else:
    print (f'{n} is not a perfect number')
```

```python
#W5A5
#Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và trả về vị trí của
#phần tử đầu tiên có giá trị k trong list số nguyên, nếu không có thì trả về -1
def index(n, k):
    for i in range (len(n)):
        if n[i] == k:
            return i
    return -1
my_list = list(map(int, input().split()))
while True:
    k = int(input("Nhap k:"))
    if k > 0:
        break
    print("k phai la so nguyen duong")
print (f' Vi tri cua phan tu dau tien co gia tri k trong list la: {index(my_list, k)}')
```

```python
#W5A6
#Viết hàm tính giai thừa. Nhập vào số nguyên n, trả về giai thừa của n (n!)
def factorial(n):
    if n < 0:
        return None
    pro = 1
    for i in range (1, n+1):
        pro *= i
    return pro
factorial(4)
```

```python
#W5A7
def calculator(a, b, k):
    if k == "-":
        result = a - b
    elif k == "+":
        result = a + b
    elif k == "*":
        result = a * b
    elif k == "/":
        if b == 0:
            return "Error: division by zero"
        result = a / b
    else:
        return "Error: invalid operator"

    return f"{result:.2f}"
num1, num2 = map(float, input(). split())
operat = input()
print(calculator(num1, num2, operat))
```

```python
#W5A8
def binary(n):
    if n == 0:
        return "0"
    bits = ""
    while n > 0:
        bits = str(n % 2) + bits
        n //= 2
    binary = 0
    for i in bits:
        binary = binary * 10 + int(i)
    return binary
def Hamming(x, y):
    x = binary(x)
    y = binary(y)
    count = 0
    larger = max (x, y)
    smaller = min (x, y)
    while smaller != 0:
        if (smaller % 10) != (larger % 10):
            count += 1
        smaller //= 10
        larger //= 10
    while larger != 0:
        if (larger % 10) == 1:
            count += 1
        larger //= 10
    return count
x, y = map(int, input("Nhap x va y:").split())
print(f'K/c Hamming: {Hamming(x, y)}')
```

```python
#W5A8
def hamming(x, y):
    diff = x ^ y
    count = 0
    while diff:
        diff &= diff - 1 #diff = diff & (diff - 1)
        count += 1
    return count
x, y = map(int, input().split())
print(f'K/c Hamming: {hamming(x, y)}')
```

```python
#W5A9
def numberofdigits(n):
    if n < 0:
        return "Not valid"
    if n == 0:
        return 1
    count = 0
    while n:
        count += 1
        n //= 10
    return count
n = int(input())
print(numberofdigits(n))
```

```python
#W5A10
def isomorphic(a, b):
    if len(a) != len(b):
        return "Invalid"
    I = {}
    for i in range (len(a)):
        ch = a[i]
        if ch not in I:
            I[ch] = b[i]
        else:
            if I[ch] != b[i]:
                return False
    for i in I:
        for j in I:
            if I[i] == I[j] and i != j:
                return False
    return True
a, b = input().split()
print(isomorphic(a, b))
```
