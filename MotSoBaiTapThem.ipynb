```python
#Bai29
def tohop (n, k):
    p = 1
    m = 1
    q = 1
    for i in range (1, n+1):
        p *= i
    for j in range (1, (n-k) + 1):
        m *= j
    for h in range (1, k+1):
        q *= h
    return int(p/(m*q))
while True:
    n = int(input("Nhap chieu cao tam giac Pascal:"))
    if n >= 0:
        break
    print ("gia tri phai >= 0")
for i in range (n):
    print (' ' * (n - i), end = '')
    for j in range (0, i):
        print (tohop(i, j), end = ' ')
    print ('1')
```

```python
#Bai31
def Fibonacci(n):
    leader = 1
    follower = 0
    if n == 1:
        return follower
    for i in range (n-2):
        leader = leader + follower
        follower = leader - follower
    return leader
while True:
    n = int(input("Nhap so thu tu cua so mong muon trong day Fibonacci:"))
    if n > 0:
        break
    print ("n phai lon hon 0")
print (Fibonacci(n))
```

```python
#Bai31
n = int(input("Nhap mot so trong day Fibonacci:"))
if n < 0:
    print("False")
else:
    check = True
    leader = 1
    follower = 0
    while (check):
        if n == follower or n == leader:
            break
        leader += follower
        follower = leader - follower
        if n < leader:
            check = False
            break
if check:
    print ("True")
else:
    print ("False")
```

```python
#Bài 33: Tính tổng S =   1/i!(với i= 0,…., n; dùng hàm con)
def tong(n):
    tong = 0
    for i in range (0, n+1):
        count = 1
        for j in range (1, i+1):
            count *= j
        tong += 1 / count
    return tong
while True:
    try:
        n = int(input("Nhap n:"))
        if n >= 0:
            break
        print ("n phai lon hon hoac bang 0")
    except ValueError:
        print ("Nhap lai")
print (tong(n))
```

```python
#Bài 35
def tong(x, n):
    tong = 0
    for i in range (0, n+1):
        tu = 1 + x**i
        mau = 1
        for j in range (1, i+2):
            mau *= j
        tong += tu / mau
    return tong
while True:
    try:
        n = int(input("Nhap n:"))
        x = float(input("Nhap x:"))
        if n >= 0:
            break
        print ("n phai lon hon hoac bang 0")
    except ValueError:
        print ("Nhap lai")
print (tong(x, n))
```
