```python
#Bai3
a, b, c = map(float, input("Nhap ba canh tam giac:").split())
if a + b > c and b + c > a and a + c > b:
    if a == b == c:
        print("Tam giac deu")
    elif a == b or b == c or a == c:
        print ("Tam giac can")
    elif a**2 + b**2 == c**2 or a**2 + c**2 == b**2 or b**2 + c**2 == a**2:
        print ("Tam giac vuong")
    else:
        print ("Tam giac thuong")
    cv = a + b + c
    p = cv / 2
    S = p*(p-a)*(p-b)*(p-c)
    S = S**0.5
    print(f'Dien tich: {S:.2f}\nChu vi: {cv:.2f}')
else: print ("Khong la ba canh cua mot tam giac")
```

```python
#Bai6
a, b, c = map(float, input("Nhap a, b, c:").split())
delta = b**2 - 4*a*c
if delta > 0:
    x1 = (-b + delta**0.5)/(2*a)
    x2 = (-b - delta**0.5)/(2*a)
    print(f'Nghiem cua phuong trinh la:\nx1 = {x1}\nx2 = {x2}')
elif delta == 0:
    x1 = (-b + delta**0.5)/(2*a)
    print(f'Nghiem cua phuong trinh la: {x1}')
else:
    print("Phuong trinh vo nghiem")
```

```python
#Bai28
def chinhhop(n,k):
    p = 1
    m = 1
    b = 1
    for i in range (1, k + 1):
        p *= i
    for j in range (1, n + 1):
        m *= j
    for l in range (1, (n-k) + 1):
        b *= l
    return m/(p*b)
while True:
    n, k = map(int, input("Nhap n va k:").split())
    if n >= k:
        print (chinhhop(n,k))
        break
    print("k be hon hoac bang n")
```
