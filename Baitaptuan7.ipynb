```python
#W7A1
# Viết hàm tìm kiếm nhị phân trên một danh sách đã được sắp xếp tăng dần.
# Trả về chỉ số của phần tử hoặc -1 nếu không tìm thấy. 
# mẫu hàm: def binary_search(arr, left, right, target):
def binary_search(arr, target):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1
arr = [int(i) for i in input().split()]
k = int(input())
print(binary_search(arr, k))
```

```python
#W7A2
# Cho danh sách arr và một số nguyên x, đếm xem x
# xuất hiện bao nhiêu lần trong danh sách. def count_occurrences(arr, x) -> int:
def count_occurences(arr, x) -> int:
    count = 0
    for i in range(len(arr)):
        if arr[i] == x:
            count += 1
    return count
arr = list(map(int, input().split()))
x = int(input())
print(count_occurences(arr, x))
```

```python
#W7A3
# Viết hàm thực hiện thuật toán bubble sort, Selection Sort để sắp xếp danh sách
# theo thứ tự tăng dần. Sắp xếp nổi bọt (Bubble Sort); Sắp xếp chọn (Selection Sort)
def Bubblesort(arr):
    n = len(arr)
    for i  in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True
        if not swapped:
            break
def Selectionsort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[min_idx], arr[i] = arr[i], arr[min_idx]

arr = list(map(int, input().split()))
Selectionsort(arr)
print(*arr, sep = ' ')
```

```python
#W7A4
# Viết hàm tìm phần tử có số lần xuất hiện nhiều nhất
# trong danh sách. Nếu có nhiều phần tử cùng số lần,
# trả về phần tử xuất hiện trước.
def most_freg(arr):
    dic = {}
    for i in range(len(arr)):
        if arr[i] not in dic:
            dic[arr[i]] = 0
        dic[arr[i]] += 1
    count = 0
    for key, value in dic.items():
        if value > count:
            count = value
            n = key
    return f"{n} xuat hien nhieu nhat, som nhat, {count} lan"
arr = list(map(int, input().split()))
print(most_freg(arr))
```

```python
#W7A5
# Cho danh sách các số nguyên và một số X, đếm
# tất cả các cặp (a, b) sao cho a + b = X.
def count_abX(arr, X):
    count = 0
    for i in range(len(arr)):
        for j in range(i+1, len(arr)):
            if arr[i] + arr[j] == X:
                count += 1
    return count
arr = list(map(int, input().split()))
X = int(input())
print(count_abX(arr, X))
```

```python
#W7A6
# Bạn được cung cấp một danh sách các danh sách
# lồng nhau, vd [[1, 2, 3], [4, 5], [6, 7, 8, 9]], hãy làm phẳng danh sách
def flat(arr):
    my_list = []
    for i in range(len(arr)):
        for j in range(len(arr[i])):
            my_list.append(arr[i][j])
    return (my_list)
arr = eval(input())
print(flat(arr))
```

```python
#W7A7
# Cho một danh sách các số nguyên nums, hãy tìm độ dài của dãy con tăng
# nghiêm ngặt dài nhất (không nhất thiết phải liền kề).
def strictly_increasing(arr):
    n = len(arr)
    dp = [1] * n
    for i in range(n):
        for j in range(i):
            if arr[j] < arr[i]:
                dp[i] = max(dp[i], dp[j] + 1)
    return max(dp)
arr = eval(input())
print(strictly_increasing(arr))
```
