```python
from pathlib import Path
WORKDIR = Path(".")

path = input("Moi ban nhap ten file:")
def reverse(path:str) -> int:
    myfile = WORKDIR/path
    with open(myfile, "r") as f:
        n = int(f.read().strip())
    n_rev = 0
    while n:
        n_rev = n_rev*10 + n%10
        n //= 10
    return n_rev
print(reverse(path))
print("Check your file!!!")
```

```python
from pathlib import Path
WORKDIR = Path(".")

path = input("Moi ban nhap ten file:")

def canWinNim(path:str) -> bool:
    file = WORKDIR/path #Day du ca duong dan va ten file
    with open(file, "r") as f:
        n = int(f.read().strip())

    if n % 4 == 0:
        return False
    else:
        return True
if (canWinNim(path)):
    print("Toi thang roi!!!")
else:
    print("Ban thang roi!!!")

print("Passed tests!")
```

```python
from pathlib import Path
WORKDIR = Path(".")
path1 = input("Moi ban nhap ten file 1:")
path2 = input("Moi ban nhap ten file 2:")
def areAnagrams(path1: str, path2: str):
    file1 = WORKDIR/path1
    file2 = WORKDIR/path2
    with open(file1, "r") as f:
        text1 = f.read().strip()
        cleaned_1 = [c.lower() for c in text1 if c.isalpha()]
    with open(file2, "r") as f:
        text2 = f.read().strip()
        cleaned_2 = [c.lower() for c in text2 if c.isalpha()]
    dic1 = {}
    dic2 = {}
    for ch in cleaned_1:
        if ch not in dic1:
            dic1[ch] = 0
        dic1[ch] += 1
    for ch in cleaned_2:
        if ch not in dic2:
            dic2[ch] = 0
        dic2[ch] += 1
    return dic1 == dic2
print(areAnagrams(path1, path2))
```

```python
from pathlib import Path
WORKDIR = Path(".")
path = input("Moi ban nhap ten file:")
character = input("Moi ban nhap character:")
def count(path: str, character: str):
    file = WORKDIR/path
    with open(file, "r") as f:
        text = f.read().strip()
    count = 0
    for i in range(len(text) - len(character) + 1):
        if text[i : i + len(character)] == character:
            count += 1
    return count
print(count(path, character))
```
