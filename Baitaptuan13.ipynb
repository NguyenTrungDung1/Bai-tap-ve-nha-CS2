```python
from pathlib import Path
WORKDIR = Path(".")

path = input("Moi ban nhap ten file:")
def reverse(path:str) -> int:
    myfile = WORKDIR/path
    with open(myfile, "r") as f:
        n = int(f.read().strip())
    n_rev = 0
    while n:
        n_rev = n_rev*10 + n%10
        n //= 10
    return n_rev
print(reverse(path))
print("Check your file!!!")
```

```python
from pathlib import Path
WORKDIR = Path(".")
path1 = input("Moi ban nhap ten file 1:")
path2 = input("Moi ban nhap ten file 2:")
def areAnagrams(path1: str, path2: str):
    file1 = WORKDIR/path1
    file2 = WORKDIR/path2
    with open(file1, "r") as f:
        text1 = f.read().strip()
        cleaned_1 = [c.lower() for c in text1 if c.isalpha()]
    with open(file2, "r") as f:
        text2 = f.read().strip()
        cleaned_2 = [c.lower() for c in text2 if c.isalpha()]
    dic1 = {}
    dic2 = {}
    for ch in cleaned_1:
        if ch not in dic1:
            dic1[ch] = 0
        dic1[ch] += 1
    for ch in cleaned_2:
        if ch not in dic2:
            dic2[ch] = 0
        dic2[ch] += 1
    return dic1 == dic2
print(areAnagrams(path1, path2))
```

```python
from pathlib import Path
WORKDIR = Path(".")
path = input("Moi ban nhap ten file:")
character = input("Moi ban nhap character:")
def count(path: str, character: str):
    file = WORKDIR/path
    with open(file, "r") as f:
        text = f.read().strip()
    count = 0
    for i in range(len(text) - len(character) + 1):
        if text[i : i + len(character)] == character:
            count += 1
    return count
print(count(path, character))
```

```python
from pathlib import Path
WORKDIR = Path(".")
path = input("Moi ban nhap ten file:")
letter = input("Moi ban nhap ki tu:")
def countCharacter(path:str, letter:str):
    file = WORKDIR/path
    with open(file, "r") as f:
        text = f.read()
        cleaned = [c.lower() for c in text if c.isalpha()]
    dic = {}
    for x in cleaned:
        if x not in dic:
            dic[x] = 0
        dic[x] += 1
    count = dic.get(letter, 0)
    return count
print(countCharacter(path, letter))
```

```python
from pathlib import Path
WORKDIR = Path(".")
path = "output.txt"
n = int(input("Nhap n:"))
def Fibonacci(n):
    a, b = "a", "b"
    for i in range(n):
        a, b = b, b + a
    return a
def list_of_fibonacci(n):
    text = ""
    for i in range (n + 1):
        text += f"f({i}) = {Fibonacci(i)}\n"
    return text.strip()
file = WORKDIR/path
with open(file, "w") as f:
    f.write(list_of_fibonacci(n))
```

```python
from pathlib import Path
WORKDIR = Path(".")
path = input("Moi ban nhap ten file:")
def getMoney(path:str) -> list[int]:
    file = WORKDIR/path
    with open(file, "r") as f:
        arr = list(map(int, f.read().split()))
    return arr
def rob(houses: list[int]) -> int:
    if not houses:
        return 0
    dp = [0] * len(houses)
    for i in range(len(houses)):
        if i == 0:
            dp[i] = houses[0]
        elif i == 1:
            dp[i] = max(dp[i - 1], houses[i])
        else:
            dp[i] = max(dp[i - 1], houses[i] + dp[i - 2])
    return dp[-1]
print(rob(getMoney(path)))
```

```python
from pathlib import Path
WORKDIR = Path(".")
path = input("Moi ban nhap ten file:")
def maximumProduct (path: str) -> int:
    n = 10
    with open(WORKDIR/path, "r") as f:
        f.seek(len(str(n)) + 1)
        nums = f.read()
    nums = list(map(int, nums.split()))
    nums.sort()
    return max(nums[-1]*nums[-2]*nums[-3], nums[0]*nums[1]*nums[-1])
print(maximumProduct(path))
```

```python
from pathlib import Path
WORKDIR = Path(".")
path = input("Moi ban nhap ten file:")
def moveZeroes (path:str) -> list:
    with open(WORKDIR/path, "r") as f:
        n = int(f.readline())
        arr = list(map(int, f.read().split()))
    i = 0
    for j in range(len(arr)):
        if arr[i] == 0:
            arr.pop(i)
            arr.append(0)
        else:
            i += 1
    return arr
print(moveZeroes(path))
```

```python
from pathlib import Path
path = input("Moi ban nhap ten file:")
WORKDIR = Path(".")
def circumference(path:str):
    with open(WORKDIR/path, "r") as f:
        text = f.readlines()
    i = 0
    while i < len(text):
        row = text[i].strip()
        if row == "":
            i += 1
            continue
        key, value = row.split(":")
        shape = value.strip()
        if shape == "SQUARE":
            i += 1
            next_row = text[i].strip().split(":")
            CV = float(next_row[1]) * 4
            print(f"{CV:.2f}")
            i += 1
        elif shape == "RECTANGLE":
            i += 1
            next_row = text[i].strip().split()
            width = next_row[1]
            height = next_row[3]
            CV = (float(width) + float(height)) * 2
            print(f"{CV:.2f}")
            i += 1
        elif shape == "CIRCLE":
            i += 1
            next_row = text[i].strip().split(":")
            CV = float(next_row[1]) * 2 * 3.14
            print(f"{CV:.2f}")
            i += 1
circumference(path)
```

```python
from pathlib import Path
WORKDIR = Path(".")
path = input("Moi ban nhap ten file:")
def productExceptSelf(path:str) -> list[int]:
    with open(WORKDIR/path, "r") as f:
        f.readline()
        arr = list(map(int, f.readline().split()))
    n = len(arr)
    res = [1] * n
    prefix = 1
    for i in range (n):
        res[i] = prefix
        prefix *= arr[i]
    suffix = 1
    for j in range(n - 1, -1, -1):
        res[j] *= suffix
        suffix *= arr[j]
    return " ".join(str(x) for x in res)
print(productExceptSelf(path))
```

```python
from pathlib import Path
import math
WORKDIR = Path(".")
path = input("Moi ban nhap ten file:")
def estimatedTime(path: str):
    with open(WORKDIR/path, "r") as f:
        src = f.readline().split()
        des = f.readline().split()
        velocity = f.readline().split()
    x1, y1 = int(src[src.index("X:") + 1]), int(src[src.index("Y:") + 1])
    x2, y2 = int(des[des.index("X:") + 1]), int(des[des.index("Y:") + 1])
    dist = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
    velocity = int(velocity[1])
    return f"{dist/velocity:.2f}"
print(estimatedTime(path))
```
